```python
# Cargar librerías necesarias
import pandas as pd
import numpy as np
```

```python
# Cargar la base de datos
dataF = pd.DataFrame(np.array([[7, 6.5, 9.2, 8.6, 8],
                            [7.5, 9.4, 7.3, 7, 7],
                            [7.6, 9.2, 8, 8, 7.5],
                            [5, 6.5, 6.5, 7, 9],
                            [6, 6, 7.8, 8.9, 7.3],
                            [7.8, 9.6, 7.7, 8, 6.5],
                            [6.3, 6.4, 8.2, 9, 7.2],
                            [7.9, 9.7, 7.5, 8, 6],
                            [6, 6, 6.5, 5.5, 8.7],
                            [6.8, 7.2, 8.7, 9, 7]]),
                  index=['Lucia', 'Pedro', 'Ines', 'Luis', 'Andres', 'Ana', 'Carlos', 'Jose', 'Sonia', 'Maria'],
                  columns=['Matematicas', 'Ciencias', 'Español', 'Historia', 'EdFisica'])
dataF
```

```python
# Extraer los nombres de las filas como etiquetas
etiquetas = dataF.index
etiquetas
```

```python
# Asignar el DataFrame original a la variable A
A= dataF
A
```

```python
# Importar librerías adicionales para SVD y visualización
from numpy import diag
from numpy import zeros
from scipy.linalg import svd
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
U, s, VT = svd(A)   # Obtención de matrices cuyo producto arroja la matriz A
Sigma = zeros((A.shape[0], A.shape[1]))  # Creación de matriz de ceros de la misma dimensión que A
Sigma[:A.shape[1], :A.shape[1]] = diag(s)
```

```python
U, s, VT = svd(A)   # Obtención de matrices cuyo producto arroja la matriz A
Sigma = zeros((A.shape[0], A.shape[1]))  # Creación de matriz de ceros de la misma dimensión que A
Sigma[:A.shape[1], :A.shape[1]] = diag(s)
```

```python
# Imprimir las matrices obtenidas mediante SVD
print("Matriz U: ")
print(U)
```

```python
print("Matriz Sigma: ")
print(Sigma)
```

```python
print("Matriz VT: ")
print(VT)
```

```python
# Reconstruir la matriz original B multiplicando las matrices U, Sigma y VT
B = U.dot(Sigma.dot(VT))
print(B)
```

```python
# Reducción a DOS Dimensiones
n_elements = 2

UReduced2 = U[:, :n_elements]
print("U Reducida:")
print(UReduced2)
```

```python
# Reducir la matriz Sigma a 2 dimensiones
SigmaReduced2 = Sigma[:n_elements, :n_elements]
print("Matriz Sigma Reducida")
print(SigmaReduced2)
```

```python
# Reducir la matriz VT a 2 dimensiones
VTReduced2 = VT[:n_elements, :]
print("VT Reducida")
print(VTReduced2)
```

```python
# Reconstrucción aproximada de la matriz A usando 2 componentes
AReduced2 = UReduced2.dot(SigmaReduced2.dot(VTReduced2))
print("Matriz A transformada:")
print(AReduced2)
```

```python
# Proyectar las observaciones a 2 dimensiones en la matriz T2
T2 = UReduced2.dot(SigmaReduced2)
print("Matriz T (2 Dimensiones):")
print(T2)
```

```python
# Visualización básica en 2D con un gráfico de dispersión
df = pd.DataFrame(T2, columns=["a", "b"])
df.plot(kind="scatter", x="a", y="b")
plt.show()
```

```python
# Graficar las observaciones en 2D anotando los nombres de los estudiantes
df = pd.DataFrame(T2, columns=["a", "b"])
x = df.iloc[:, 0]
y = df.iloc[:, 1]
x = x.to_numpy()
# y = y.to_numpy()

fig, ax = plt.subplots()
ax.set_title("Mapa de observaciones")
ax.scatter(x, y)

for i, txt in enumerate(etiquetas):
    ax.annotate(txt, (x[i], y[i]))
```

```python
# Reducción a 3 Dimensiones
n_elements = 3

UReduced3 = U[:, :n_elements]

SigmaReduced3 = Sigma[:n_elements, :n_elements]

VTReduced3 = VT[:n_elements, :]

AReduced3 = UReduced3.dot(SigmaReduced3.dot(VTReduced3))
print("Matriz A transformada:")
print(AReduced3)
```

```python
# Proyectar las observaciones a 3 dimensiones en la matriz T3
T3 = UReduced3.dot(SigmaReduced3)
print("Matriz T (3 Dimensiones):")
print(T3)
```

```python
# Importar módulo para gráficos 3D
from mpl_toolkits import mplot3d
%matplotlib inline
```

```python
# Convertir la matriz reducida de 3 dimensiones en DataFrame
df3 = pd.DataFrame(T3, columns = ["a", "b", "c"])
df3
```

```python
# Generar un gráfico de dispersión en 3D
fig = plt.figure()
ax = plt.axes(projection="3d")
xline = df3["b"]
yline = df3["c"]
zline = df3["a"]
ax.scatter3D(xline, yline, zline)
```

```python
# Graficar observaciones en 3D etiquetando cada punto con su nombre
from mpl_toolkits.mplot3d import Axes3D


fig = plt.figure()
ax3d = fig.add_subplot(projection="3d")

arrayx = np.array(df3["a"])
arrayy = np.array(df3["b"])
arrayz = np.array(df3["c"])

ax3d.scatter(arrayx, arrayy, arrayz)

for x, y, z, label in zip(arrayx, arrayy, arrayz, etiquetas):
        ax3d.text(x, y, z, label)

plt.title("Datos en 3D")
plt.show()
```

```python
# Calcular y graficar el error de reconstrucción en diagrama de caja (2D)
plt.figure(figsize = (10,7))
Error2 = B - AReduced2

dfError2 = pd.DataFrame(Error2, columns = ["Matematicas", "Ciencias", "Español", "Historia", "EdFisica"])
dfError2.boxplot()
```

```python
# Mostrar el promedio de error por materia (2D)
print("Errores promedio por columna")
Error2.mean(axis=0)
```

```python
# Calcular y graficar el error de reconstrucción en diagrama de caja (3D)
Error3 = B - AReduced3

plt.figure(figsize = (10, 7))
dfError3 = pd.DataFrame(Error3, columns =["Matematicas", "Ciencias", "Español", "Historia", "EdFisica"])
dfError3.boxplot()
plt.show()
```

```python
# Mostrar el promedio de error por materia (3D)
print("Errores promedio por columna")
Error3.mean(axis=0)
```
